## BertBackbone für Feature Extraktion

```python
preprocessor = keras_hub.models.BertPreprocessor.from_preset(bert_name, sequence_length=512)
backbone = keras_hub.models.BertBackbone.from_preset(bert_name)

inp = preprocessor(["hallo", "this is a test"])
print(inp.keys())
inp['token_ids']
```

```python
backbone.trainable = False

input = tf.keras.Input(shape=(), dtype=tf.string)
output = backbone(preprocessor(input))
embedding = output['pooled_output']

x = tf.keras.layers.Dense(128, activation='selu')(embedding)
x = tf.keras.layers.Dense(128, activation='selu')(x)
x = tf.keras.layers.Dense(64, activation='relu')(x)
x = tf.keras.layers.Dense(16, activation='selu')(x)
y = tf.keras.layers.Dense(1, activation='sigmoid')(x)

model = tf.keras.Model(input, y)
model.summary()
```

```python
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
        patience=2,
        monitor='val_loss',
        min_delta=0.01,
        restore_best_weights=True
        )
model.fit(train_ds, validation_data=val_ds, epochs=10, callbacks=[early_stopping], verbose=1)
```

```python
inp_tf = tf.constant(["What an amazing movie!", "A total waste of my time."])
model.predict(inp_tf)
```

## BertClassifier: Eigene Implementierung

```python
class BertClassifier:
    def __init__(self, name, layers, nr_of_classes, activations="ReLU", output_activation="softmax", dropout=None,
                 sequence_length=512
                 ):
        self.name = name
        self.hidden_units = self._ensure_list(layers, "layers")
        self.activations = self._ensure_list(activations, "activations", expected_len=len(self.hidden_units))
        self.output_activation = output_activation
        self.dropouts = self._ensure_list(dropout, "dropout", expected_len=len(self.hidden_units), allow_none=True)
        self.nr_of_classes = nr_of_classes
        self.sequence_length = sequence_length
        self.model = self._build_model()

    @staticmethod
    def _ensure_list(value, argument_name, expected_len=None, allow_none=False):
        if value is None and allow_none:
            values = [None] * expected_len
        elif isinstance(value, (list, tuple)):
            values = list(value)
        else:
            values = [value]

        if expected_len is not None:
            if len(values) == 1 and expected_len != 1 and argument_name != "layers":
                values = values * expected_len
            elif len(values) != expected_len:
                raise ValueError(
                        f"{argument_name} muss entweder ein einzelner Wert sein oder "
                        f"genau {expected_len} Werte enthalten."
                        )
        return values

    @staticmethod
    def _normalize_activation(activation):
        if isinstance(activation, str):
            return activation.lower()
        return activation

    def _build_model(self):
        preprocessor = keras_hub.models.BertPreprocessor.from_preset(
                self.name,
                sequence_length=self.sequence_length,
                )
        backbone = keras_hub.models.BertBackbone.from_preset(self.name)
        backbone.trainable = False

        input_layer = tf.keras.Input(shape=(), dtype=tf.string)
        output = backbone(preprocessor(input_layer))
        x = output["pooled_output"]

        for units, activation, dropout in zip(self.hidden_units, self.activations, self.dropouts):
            x = tf.keras.layers.Dense(
                    units,
                    activation=self._normalize_activation(activation),
                    )(x)
            if dropout is not None:
                x = tf.keras.layers.Dropout(dropout)(x)

        y = tf.keras.layers.Dense(self.nr_of_classes, activation=self.output_activation)(x)
        return tf.keras.Model(input_layer, y, name="bert_classifier")

    def __call__(self, *args, **kwargs):
        return self.model(*args, **kwargs)

    def __getattr__(self, item):
        model = self.__dict__.get("model")
        if model is None:
            raise AttributeError(item)
        return getattr(model, item)
```

```python
custom_bert_classifier = BertClassifier(
        bert_name,
        layers=[128, 128, 64, 16],
        nr_of_classes=2,
        activations=["selu", "selu", "relu", "selu"],
        sequence_length=512,
        )
custom_bert_classifier.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
custom_bert_classifier.summary()
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
        patience=2,
        monitor='val_loss',
        min_delta=0.01,
        restore_best_weights=True
        )
custom_bert_classifier.fit(train_ds, validation_data=val_ds, epochs=10, callbacks=[early_stopping], verbose=1)
```

```python
custom_bert_classifier.evaluate(val_ds_full)
```

```python
inp_tf = tf.constant(["What an amazing movie!", "A total waste of my time."])
custom_bert_classifier.predict(inp_tf)
```
